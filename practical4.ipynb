```python

N = 8

# Create an empty 8x8 chessboard
board = [[0 for i in range(N)] for j in range(N)]

# Initial position of first queen (0-based indexing)
x = 3
y = 2

# Place the first queen
board[x][y] = 1


def is_safe(row, col):

    for i in range(N):
        for j in range(N):

            if board[i][j] == 1:

                if i == row and j == col:
                    continue

                # Check same row or column
                if i == row or j == col:
                    return False

                # Check diagonal
                if abs(i - row) == abs(j - col):
                    return False

    return True


def solve(col):

    if col >= N:
        return True

    # Skip the column containing the first queen
    if col == y:
        return solve(col + 1)

    for row in range(N):

        if row == x and col != y:
            continue

        if is_safe(row, col):

            board[row][col] = 1

            if solve(col + 1):
                return True

            # Backtracking
            board[row][col] = 0

    return False


if solve(0):

    print("N Queen Backtracking Solution:")
    print("Initial position of first queen:", x, y)

    for row in board:
        print(" ".join(map(str, row)))

else:
    print("Solution does not exist")
```
