```python
n = int(input("Enter number of jobs: "))

jobs = []

for i in range(n):
    deadline = int(input(f"Enter deadline for Job {i+1}: "))
    profit = int(input(f"Enter profit for Job {i+1}: "))
    jobs.append([i+1, deadline, profit])

# Sort jobs by profit in descending order
jobs.sort(key=lambda x: x[2], reverse=True)

max_deadline = max(job[1] for job in jobs)

slots = [False] * (max_deadline + 1)
result = [None] * (max_deadline + 1)

total_profit = 0

# Schedule jobs using greedy method
for job in jobs:
    for j in range(min(job[1], max_deadline), 0, -1):
        if not slots[j]:
            slots[j] = True
            result[j] = job[0]
            total_profit += job[2]
            break

print("\nScheduled Jobs:")

for j in range(1, max_deadline + 1):
    if result[j] is not None:
        print("Job", result[j], "in slot", j)

print("Maximum Profit =", total_profit)
```
