```python
from sympy import Symbol, lambdify
import matplotlib.pyplot as plt
import numpy as np

# Define variable
x = Symbol('x')

# Define the function
function = (x + 3)**2

def gradient_descent(function, start, learn_rate,
                     n_iter=10000, tolerance=1e-6):

    # Calculate derivative
    gradient = lambdify(x, function.diff(x))
    func = lambdify(x, function)

    points = [start]
    step_size = 1
    iterations = 0

    # Gradient Descent
    while step_size > tolerance and iterations < n_iter:
        previous_x = start

        start = start - learn_rate * gradient(previous_x)

        step_size = abs(start - previous_x)
        iterations += 1
        points.append(start)

    print("The local minimum occurs at:", start)
    print("Minimum value of function:", func(start))
    print("Number of iterations:", iterations)

    # Plot the function
    x_values = np.linspace(-7, 5, 100)
    y_values = func(x_values)

    plt.figure(figsize=(8, 5))
    plt.plot(x_values, y_values, label="y = (x+3)^2")
    plt.plot(points, func(np.array(points)), '-o',
             color='red', label="Gradient Descent Steps")

    plt.xlabel("x")
    plt.ylabel("y")
    plt.title("Gradient Descent Algorithm")
    plt.legend()
    plt.grid()
    plt.show()


# Starting point and learning rate
gradient_descent(
    function=function,
    start=2.0,
    learn_rate=0.2,
    n_iter=50
)
```
