```python

import threading
import time
import random

# Function to merge two sorted arrays
def merge(left, right):
    result = []
    i = j = 0

    while i < len(left) and j < len(right):
        if left[i] < right[j]:
            result.append(left[i])
            i += 1
        else:
            result.append(right[j])
            j += 1

    result.extend(left[i:])
    result.extend(right[j:])

    return result


# Normal Merge Sort
def merge_sort(arr):
    if len(arr) <= 1:
        return arr

    mid = len(arr) // 2

    left = merge_sort(arr[:mid])
    right = merge_sort(arr[mid:])

    return merge(left, right)


# Multithreaded Merge Sort
def multithreaded_merge_sort(arr):
    if len(arr) <= 1:
        return arr

    mid = len(arr) // 2

    left = arr[:mid]
    right = arr[mid:]

    # Create two threads
    t1 = threading.Thread(
        target=left.extend,
        args=(merge_sort(left),)
    )

    t2 = threading.Thread(
        target=right.extend,
        args=(merge_sort(right),)
    )

    # Start threads
    t1.start()
    t2.start()

    # Wait for both threads
    t1.join()
    t2.join()

    return merge(left, right)


# Generate input array
arr = [random.randint(1, 10000) for i in range(20000)]

# Normal Merge Sort
a = arr.copy()

start = time.perf_counter()
sorted_a = merge_sort(a)
normal_time = time.perf_counter() - start

# Multithreaded Merge Sort
b = arr.copy()

start = time.perf_counter()
sorted_b = multithreaded_merge_sort(b)
multi_time = time.perf_counter() - start

# Display results
print("Normal Merge Sort Time:", normal_time)
print("Multithreaded Merge Sort Time:", multi_time)

print("Normal Sorted:", sorted_a[:10])
print("Multithreaded Sorted:", sorted_b[:10])

print("Both results are correct:", sorted_a == sorted_b)
```
